# Exp A — Is an event a threat? (Def 5 precondition)
Semantic-only novelty relative to a model of normal (benign) behaviour. Reports detection metrics (§6.1). Note: semantic-only by design — `A`/`σ` are label-derived and would make this circular.

```
import sys; sys.path.insert(0, '../src')
import matplotlib.pyplot as plt
from zeroday_verify import data as D
from zeroday_verify.experiments import exp_a_threat_vs_normal
paths = D.discover_csvs('../data/raw')
df = D.load_subsampled(paths, per_attack_family=400, n_benign=2000, seed=17)
threats = D.build_threats(df)
a = exp_a_threat_vs_normal(threats)
print({k: round(v,3) for k,v in a['metrics'].items()})
print('raw-feature reference AUROC:', round(a['raw_baseline_auroc'],3))
print('per-family mean novelty u:', {k: round(v,3) for k,v in a['per_family_mean_u'].items()})
```
